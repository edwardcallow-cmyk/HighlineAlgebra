## Phase 1 — One point (by hand first) / Un punto (primero a mano)

A 2×2 matrix looks like:
\[
A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}
\]
A point can be written as a column vector:
\[
v = \begin{pmatrix} x \\ y \end{pmatrix}
\]
Then:
\[
Av = \begin{pmatrix} a & b \\ c & d \end{pmatrix}\begin{pmatrix} x \\ y \end{pmatrix}
= \begin{pmatrix} ax + by \\ cx + dy \end{pmatrix}
\]

## Phase 2 — Many points make a shape / Muchos puntos forman una figura

A shape is just a list of points. If a matrix transforms one point, it can transform **every** point in a shape.  
Una figura es solo una lista de puntos. Si una matriz transforma un punto, puede transformar **todos** los puntos de una figura.

We will use 4 shapes: **square, triangle, house, star**.  
Usaremos 4 figuras: **cuadrado, triángulo, casa, estrella**.

```
def close_shape(P):
    """Close a polygon by repeating the first point at the end."""
    if len(P) == 0:
        return P
    if not np.allclose(P[0], P[-1]):
        return np.vstack([P, P[0]])
    return P

# Define shapes as N×2 arrays (each row is [x, y]).
square = np.array([
    [-1, -1],
    [ 1, -1],
    [ 1,  1],
    [-1,  1],
], dtype=float)

triangle = np.array([
    [0,  1.6],
    [-1.4, -1.2],
    [1.4, -1.2],
], dtype=float)

house = np.array([
    [-1.6, -1.4],  # bottom-left
    [ 1.6, -1.4],  # bottom-right
    [ 1.6,  0.6],  # top-right (walls)
    [ 0.0,  1.8],  # roof peak
    [-1.6,  0.6],  # top-left (walls)
], dtype=float)

# 5-point star (simple polygon order that draws a star)
# Outer radius R, inner radius r
R, r = 1.8, 0.75
angles_outer = np.deg2rad([90, 18, -54, -126, -198])  # 5 outer points
outer = np.c_[R*np.cos(angles_outer), R*np.sin(angles_outer)]

angles_inner = np.deg2rad([54, -18, -90, -162, 126])  # 5 inner points
inner = np.c_[r*np.cos(angles_inner), r*np.sin(angles_inner)]

# Interleave outer and inner points
star = np.empty((10, 2), dtype=float)
star[0::2] = outer
star[1::2] = inner

shapes = {
    "square / cuadrado": close_shape(square),
    "triangle / triángulo": close_shape(triangle),
    "house / casa": close_shape(house),
    "star / estrella": close_shape(star),
}

def transform_points(A, P):
    """Transform an N×2 point list P by 2×2 matrix A."""
    return (A @ P.T).T

def plot_shapes(original, transformed, title_left="Original", title_right="Transformed"):
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))
    for a, P, ttl in [(ax[0], original, title_left), (ax[1], transformed, title_right)]:
        a.plot(P[:,0], P[:,1], marker="o")
        a.axhline(0, linewidth=1)
        a.axvline(0, linewidth=1)
        a.set_aspect("equal", adjustable="box")
        a.set_title(ttl)
        a.set_xlabel("x")
        a.set_ylabel("y")
        a.grid(True, linewidth=0.5)
    plt.tight_layout()
    plt.show()

# Quick preview of all shapes (original)
for name, P in shapes.items():
    plot_shapes(P, P, title_left=name, title_right=name)
```

## Mini‑Gallery (Option C) / Mini‑Galería (Opción C)

Now we will vary **one parameter** (shear \(s\)) and compare results.  
Ahora vamos a variar **un parámetro** (inclinación \(s\)) y comparar resultados.

Matrix:
\[
A_s = \begin{pmatrix}1 & s\\0 & 1\end{pmatrix}
\]
Try to notice what changes and what stays the same.  
Trata de notar qué cambia y qué se queda igual.

```
shape_name = "house / casa"
P = shapes[shape_name]

s_values = [-2, -1, 0, 1, 2]

fig, axes = plt.subplots(1, len(s_values), figsize=(15, 3))
for ax, s in zip(axes, s_values):
    A = np.array([[1, s],
                  [0, 1]], dtype=float)
    P2 = transform_points(A, P)
    ax.plot(P2[:,0], P2[:,1], marker="o")
    ax.axhline(0, linewidth=1)
    ax.axvline(0, linewidth=1)
    ax.set_aspect("equal", adjustable="box")
    ax.set_title(f"s = {s}")
    ax.grid(True, linewidth=0.5)
plt.suptitle(f"Shear gallery / Galería de inclinación — {shape_name}", y=1.05)
plt.tight_layout()
plt.show()
```
