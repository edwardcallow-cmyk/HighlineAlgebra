## Part A — When a regular graph hides the story

A **linear scale** counts by *equal adds* (…, 10, 20, 30, 40 …).  
A **log scale** counts by *equal factors* (…, 10, 100, 1000 …).

In real life, some quantities change by **multiplying**, not just adding.

### Paper Questions (Part A)
1. In your own words, what is the difference between “adding the same amount each step” and “multiplying by the same factor each step”?  
2. Which one sounds more like **exponential growth**? Why?  
3. Predict: If data grows by multiplying, what might a normal graph make hard to see?


```
# Toy data: bacteria doubling (exponential growth)
t = np.arange(0, 11)          # time steps 0 to 10
pop = 100 * (2 ** t)          # start at 100, double each step

df_toy = pd.DataFrame({"time_step": t, "population": pop})
df_toy
```

```
# Plot 1: Linear scale (normal y-axis)
plt.figure(figsize=(7,4))
plt.plot(df_toy["time_step"], df_toy["population"], marker="o")
plt.title("Toy Data: Doubling Population (Linear Scale)")
plt.xlabel("Time step")
plt.ylabel("Population")
plt.grid(True)
plt.show()
```

```
# Plot 2: Log-scaled y-axis (same data, different lens)
plt.figure(figsize=(7,4))
plt.plot(df_toy["time_step"], df_toy["population"], marker="o")
plt.yscale("log")
plt.title("Toy Data: Doubling Population (Log Scale on Y)")
plt.xlabel("Time step")
plt.ylabel("Population (log scale)")
plt.grid(True, which="both")
plt.show()
```

## Part B — Public health example (vaccination and measles)

We are going to look at a dataset that connects:
- **Vaccination coverage** (how many people are vaccinated)
- **Measles cases** (how many cases occur)

### Why this is a good log-scale example
Measles cases can vary from *near zero* to *very large* across contexts. A normal (linear) y-axis can squash the small-but-important values.

**Data source (Our World in Data):** measles vaccine coverage vs measles cases worldwide.


```
# Try to load the dataset from Our World in Data (OWID).
# If your device has no internet, the fallback sample will load instead.

url = "https://ourworldindata.org/grapher/measles-coverage-vs-cases-worldwide.csv"

try:
    df = pd.read_csv(url)
    source_used = "OWID live CSV"
except Exception as e:
    # Fallback: small sample dataset (made from the same variable names)
    from io import StringIO
    sample = StringIO("""Entity,Year,measles_cases_per_100000_people,measles_coverage
World,1980,70,16
World,1990,35,71
World,2000,20,72
World,2010,10,84
World,2019,15,85
World,2021,9,81
""")
    df = pd.read_csv(sample)
    source_used = "Fallback sample (offline)"
    
source_used, df.head()
```

```
# We'll use the "World" series to keep Day 1 simple and reduce cognitive load.
# (Later, your full project can compare multiple countries.)

if "Entity" in df.columns:
    df_world = df[df["Entity"] == "World"].copy()
else:
    df_world = df.copy()

# Keep only needed columns (names may vary slightly if OWID updates)
cols = [c for c in df_world.columns if "measles" in c.lower() or c.lower() in ["year","entity"]]
df_world = df_world[cols].dropna()

df_world.tail()
```

```
# Plot 3: Measles cases vs year (linear y-axis)
# Find the right cases column automatically
cases_col = [c for c in df_world.columns if "cases" in c.lower()][0]
year_col = [c for c in df_world.columns if c.lower() == "year"][0]

plt.figure(figsize=(8,4))
plt.plot(df_world[year_col], df_world[cases_col], marker="o")
plt.title("Measles cases (World) — Linear scale")
plt.xlabel("Year")
plt.ylabel("Measles cases per 100,000 people")
plt.grid(True)
plt.show()
```

```
# Plot 4: Same data, log-scaled y-axis
plt.figure(figsize=(8,4))
plt.plot(df_world[year_col], df_world[cases_col], marker="o")
plt.yscale("log")
plt.title("Measles cases (World) — Log scale on Y")
plt.xlabel("Year")
plt.ylabel("Measles cases per 100,000 (log scale)")
plt.grid(True, which="both")
plt.show()
```
