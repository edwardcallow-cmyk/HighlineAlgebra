## Our Quadratic Revenue Model / Nuestro modelo cuadrático de ingreso

We will use a **simple quadratic model** for daily revenue:

$$
R(p) = a \cdot p \cdot (C - p)
$$

Where:

- \( p \) = toll **price** in dollars (x-axis)
- \( C \) = **choke price** (the price where basically no one uses the road)
- \( a \) = **scale factor** that controls how big the revenue is

This model has:

- **x-intercepts** at \( p = 0 \) and \( p = C \)
- A **maximum** at the **vertex**, halfway between the intercepts: \( p = \frac{C}{2} \)

---

### En español

Vamos a usar un **modelo cuadrático sencillo** para el ingreso diario:

$$
R(p) = a \cdot p \cdot (C - p)
$$

Donde:

- \( p \) = **precio** del peaje en dólares (eje x)
- \( C \) = **precio máximo** (choke price), el precio donde básicamente nadie usa la carretera
- \( a \) = **factor de escala** que controla qué tan grande es el ingreso

Este modelo tiene:

- **intersecciones con el eje x** en \( p = 0 \) y \( p = C \)
- Un **máximo** en el **vértice**, a la mitad de las intersecciones: \( p = \frac{C}{2} \)


```python
# ============================
# Student Parameters
# ============================
# 1. Choose a choke price C (high toll where usage basically goes to 0)
#    Example idea: for a small local shortcut lane, maybe C = 10 or C = 15
C = 10.0   # <-- CHANGE THIS VALUE to your chosen choke price

# 2. Choose a scale factor a.
#    This controls how "tall" the revenue curve is.
#    Start with a small number like 100 or 200 and adjust.
a = 150.0  # <-- CHANGE THIS VALUE to your chosen scale factor

# 3. Choose a range of toll prices to explore (from 0 to C)
num_points = 100
prices = np.linspace(0, C, num_points)

def revenue(p):
    """Quadratic revenue model R(p) = a * p * (C - p)."""
    return a * p * (C - p)

revenues = revenue(prices)

C, a
```

## Comparing Different Models / Comparando Diferentes Modelos

```python
# Plot the revenue curve

plt.figure()
plt.plot(prices, revenues, marker='.')
plt.title("Revenue vs Toll Price\nIngreso vs Precio del Peaje")
plt.xlabel("Toll price p (dollars) / Precio del peaje p (dólares)")
plt.ylabel("Daily revenue R(p) / Ingreso diario R(p)")

# Mark intercepts and vertex
# Intercepts at p = 0 and p = C
plt.scatter([0, C], [0, 0], color='red')
plt.text(0, 0, "  (0, 0)", color='red')
plt.text(C, 0, f"  ({C:.1f}, 0)", color='red')

# Vertex at p = C/2
vertex_p = C / 2
vertex_R = revenue(vertex_p)
plt.scatter([vertex_p], [vertex_R], color='green')
plt.text(vertex_p, vertex_R, f"  Vertex ({vertex_p:.2f}, {vertex_R:,.0f})", color='green')

plt.show()

vertex_p, vertex_R
```

```python
# Show a small table of selected prices and revenues

sample_prices = np.linspace(0, C, 6)  # 6 evenly spaced prices from 0 to C
sample_revenues = revenue(sample_prices)

print("Price (p)\tRevenue R(p)")
for p, r in zip(sample_prices, sample_revenues):
    print(f"${p:5.2f}\t\t${r:,.2f}")
```
